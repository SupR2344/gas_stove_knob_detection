```python
import zipfile
import os

# Path to the ZIP file in your Drive or Colab environment
zip_path = "/content/drive/MyDrive/gas_stove_dataset.zip"
# Target directory where the dataset will be extracted
extract_path = "/content/data/Drive"

# Create the target directory if it doesn't exist
os.makedirs(extract_path, exist_ok=True)

# Unzip the dataset
with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully.")
```

```python
# Install YOLOv9 (Ultralytics)
!pip install ultralytics --upgrade
from ultralytics import YOLO

# This will download the official YOLOv9s model if not already present

model = YOLO('yolov9s.pt')
```
